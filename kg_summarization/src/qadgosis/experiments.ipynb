```python
import numpy as np
from sklearn.metrics import ndcg_score

# Example dictionaries (replace with your actual data)
relevance = {'item0': '3', 'item1': '2', 'item2': '3', 'item3': '1'}  # Strings
predicted_scores = {'item0': '0.9', 'item1': '0.7', 'item2': '0.8', 'item3': '0.6'}  # Strings

# Convert dictionary keys/values to numeric lists
y_true = np.array(list(relevance.values()), dtype=float)  # Convert to numeric
y_score = np.array(list(predicted_scores.values()), dtype=float)  # Convert to numeric

# Reshape for scikit-learn's ndcg_score
y_true = y_true.reshape(1, -1)
y_score = y_score.reshape(1, -1)

# Compute NDCG
ndcg = ndcg_score(y_true, y_score)
print("NDCG Score:", ndcg)
```

```python
# import required package 
from sklearn.metrics import ndcg_score, dcg_score
import numpy as np

# Relevance scores in Ideal order
true_relevance = np.asarray([[3, 2, 1, 0, 0]])
print(true_relevance)

# Relevance scores in output order
relevance_score = np.asarray([[3, 2, 0, 0, 1]])

# DCG score
dcg = dcg_score(true_relevance, relevance_score)
print("DCG score : ", dcg)

# IDCG score
idcg = dcg_score(true_relevance, true_relevance)
print("IDCG score : ", idcg)

# Normalized DCG score
ndcg = dcg / idcg
print("nDCG score : ", ndcg)

# or we can use the scikit-learn ndcg_score package
print("nDCG score (from function) : ", ndcg_score(
	true_relevance, relevance_score))
```

```python
import numpy as np
from sklearn.metrics import precision_score

# Example one-hot encoded labels
power_ranking = np.array([[1, 0, 0], [0, 1, 0], [0, 0, 1]])
monte_carlo_ranking = np.array([[1, 0, 0], [0, 0, 1], [0, 1, 0]])

# Convert to 1D class labels
power_ranking_1d = np.argmax(power_ranking, axis=1)
monte_carlo_ranking_1d = np.argmax(monte_carlo_ranking, axis=1)

# Compute precision
precision = precision_score(monte_carlo_ranking_1d, power_ranking_1d,  average='macro')
print(f"Precision: {precision}")
```
