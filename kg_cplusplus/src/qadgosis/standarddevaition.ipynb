```python
import os
import re
import pandas as pd
import gzip

# Path to the folder containing your snapshots
folder = "/Users/au735195/Desktop/wiki/snapshots"

# Pattern to match filenames like snapshot_2024-05-09_16-35-20.csv
pattern = re.compile(r"snapshot_(\d{4}-\d{2}-\d{2}_\d{2}-\d{2}-\d{2})\.csv")

# Collect and sort files by timestamp in filename
files = []
for filename in os.listdir(folder):
    match = pattern.match(filename)
    if match:
        timestamp = match.group(1)
        files.append((timestamp, filename))

# Sort by timestamp
files.sort(key=lambda x: x[0])

# Function to convert IDs to integers
def extract_int(x):
    return int(''.join(filter(str.isdigit, str(x)))) if pd.notna(x) else None

# Process each file
for i, (timestamp, filename) in enumerate(files):
    infile = os.path.join(folder, filename)
    outfile = os.path.join(folder, f"snapshot_{i}.nt.gz")

    print(f"Processing {filename} → {outfile}")

    # Read only the first three columns
    try:
        df = pd.read_csv(infile, sep="\t", usecols=["head_id", "relation_id", "tail_id"], dtype=str)
    except Exception as e:
        print(f"❌ Skipping {filename}: {e}")
        continue

    # Convert to integers
    df = df.applymap(extract_int)

    # Write as N-Triples compressed file
    with gzip.open(outfile, "wt", encoding="utf-8") as f_out:
        for _, row in df.iterrows():
            f_out.write(f"{row['head_id']} {row['relation_id']} {row['tail_id']} .\n")

print("✅ All files converted and renamed sequentially.")
```
